```python
# Import and Set Up Hugging Face Pipeline
```

```python
from transformers import pipeline

# Load a basic model for generation
generator = pipeline("text-generation", model="gpt2")
```

```python
# Run Basic Text Generation
```

```python
prompt = "The future of AI in education is"
outputs = generator(prompt, max_length=50, num_return_sequences=1)

print("Generated Text:\n", outputs[0]["generated_text"])
```

```python
# Log the run with Weights and Biases
```

```python
import wandb

wandb.init(project="llm_colab_demo", name="gpt2_prompt_test")

wandb.config.prompt = prompt
wandb.config.max_length = 50

# Log output
wandb.log({"generated_text": outputs[0]["generated_text"]})
wandb.finish()
```
